```python
#ordering countries chronologically
df=df.sort_values(by=['country','year']).reset_index(drop=True)
```

```python
df.head()
```

```python
#droping rows where critical information is missing
df=df.dropna(subset=['iso_code'])
```

```python
df.head()
```

```python
energy_cols=[col for col in df.columns if 'biofuel'in col or 'solar'in col]
df.loc[df['year']<1980,energy_cols]=df.loc[df['year']<1980,energy_cols].fillna(0.0)
```

```python
df=df.sort_values(by=['country','year']).reset_index(drop=True)
```

```python
df[energy_cols]=df.groupby('country')[energy_cols].ffill()
```

```python
#clearing out rows with no data at all

df=df.dropna(subset=['biofuel_consumption','solar_share_elec'],how='all')
```

```python
core_metrics=[col for col in df.columns if col not in ['country','year','iso_code']]

df[core_metrics]=df.groupby('country')[core_metrics].ffill()
```

```python
df[core_metrics]=df.groupby('country')[core_metrics].bfill()
```

```python
df=df.dropna(axis=1,how='all')
print(f"Remaining missing values:{df.isnull().sum().sum()}")
df.head()
```

```python
change_and_share_cols=[col for col in df.columns if any(x in col for x in['_change_','_pct','share'])]
df[change_and_share_cols]=df[change_and_share_cols].fillna(0.0)
```

```python
numeric_cols=df.select_dtypes(include=['number']).columns
```

```python
for col in numeric_cols:
    df[col]=df.groupby('year') \
            [col].transform(lambda x:
x.fillna(x.mean()))
```

```python
df=df.fillna(0.0)
print(f"Remaining missing values:{df.isnull().sum().sum()}")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib as mpl
sns.set_theme(style='whitegrid')
mpl.rcParams['figure.dpi']=100
```

```python
global_trends=df.groupby('year')[['solar_share_elec','wind_share_elec']].sum().reset_index()
global_modern=global_trends[global_trends['year']>=1990]
```

```python
import matplotlib.pyplot as plt

plt.plot(global_modern['year'],
         global_modern['solar_share_elec'],label='solar_share')
plt.plot(global_modern['year'],
         global_modern['wind_share_elec'],label='wind_share')


plt.title(
    "combined growth of solar and wind electricity share since 1990",
    fontsize=14,
    fontweight="bold",
    pad=15,
)
plt.xlabel("year",fontsize=12)
plt.ylabel("cumulative share(%)",fontsize=12)
plt.legend(loc="upper left",title="Energy source")
plt.xlim(global_modern["year"].min(),
global_modern["year"].max())

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Filter for specific milestone years
milestone_years = [1990, 2000, 2010, 2020]
milestones = global_modern[global_modern['year'].isin(milestone_years)]

# Bar chart setup
x = milestones['year']
width = 3.5  # Width of the bars

plt.figure(figsize=(10, 6))
plt.bar(x - width/2, milestones['solar_share_elec'], width, label='Solar Share', color='#f1c40f')
plt.bar(x + width/2, milestones['wind_share_elec'], width, label='Wind Share', color='#3498db')

# Formatting
plt.title("Comparison of Solar & Wind Shares across Milestone Years", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Year", fontsize=12)
plt.ylabel("Electricity Share (%)", fontsize=12)
plt.xticks(milestone_years)
plt.legend(title="Energy Source")
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax1 = plt.subplots(figsize=(10, 6))

# Plot Solar on the first y-axis (Left)
color = '#e67e22'
ax1.set_xlabel('Year', fontsize=12)
ax1.set_ylabel('Solar Share (%)', color=color, fontsize=12)
ax1.plot(global_modern['year'], global_modern['solar_share_elec'], color=color, linewidth=2, label='Solar')
ax1.tick_params(axis='y', labelcolor=color)

# Create a second y-axis sharing the same x-axis
ax2 = ax1.twinx()  
color = '#2980b9'
ax2.set_ylabel('Wind Share (%)', color=color, fontsize=12)
ax2.plot(global_modern['year'], global_modern['wind_share_elec'], color=color, linewidth=2, linestyle='--', label='Wind')
ax2.tick_params(axis='y', labelcolor=color)

plt.title("Solar vs. Wind Electricity Share Trends (Dual Y-Axis)", fontsize=14, fontweight='bold', pad=15)
fig.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Setup the data
years = global_modern['year']
solar = global_modern['solar_share_elec']
wind = global_modern['wind_share_elec']

# Create the stacked area chart
plt.figure(figsize=(10, 6))
plt.stackplot(years, solar, wind, labels=['Solar Share', 'Wind Share'], colors=['#ffcc00', '#3399ff'], alpha=0.8)

# Formatting
plt.title("Growth of Combined Solar and Wind Share in Electricity Generation", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Year", fontsize=12)
plt.ylabel("Electricity Share (%)", fontsize=12)
plt.legend(loc='upper left', title="Energy Source")
plt.xlim(years.min(), years.max())
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

# Year-over-Year (YOY)  Solar_energy growth rate


```python
duckdb.sql("""
with annual_solar as(
select country,year,solar_consumption as current_solar
from df
where Iso_code is not null and solar_consumption is not null  and year>=2015)
select t1.country,
t1.year as current_year,
round(t1.current_solar,2)as solar_this_year,
round(t2.current_solar,2) as solar_last_year,
round(((t1.current_solar-t2.current_solar)/t2.current_solar)*100,2)as YOY_solar_growth_percentage
from annual_solar t1
join annual_solar t2
    on t1.country=t2.country
    and t1.year=t2.year+1
    where t1.country='India'
    order by current_year DESC;
    """).show()
```

# Year-over-Year (YOY)  wind_energy growth rate


```python
duckdb.sql("""
with annual_wind as(
select country,year,wind_consumption as current_wind
from df
where Iso_code is not null and wind_consumption is not null  and year>=2015)
select t1.country,
t1.year as current_year,
round(t1.current_wind,2)as wind_this_year,
round(t2.current_wind,2) as wind_last_year,
round(((t1.current_wind-t2.current_wind)/t2.current_wind)*100,2)as YOY_wind_growth_percentage
from annual_wind t1
join annual_wind t2
    on t1.country=t2.country
    and t1.year=t2.year+1
    where t1.country='India'
    order by current_year DESC;
    """).show()
```

# Year-over-Year (YOY)  biofuel growth rate


```python

duckdb.sql("""
with annual_biofuel as(
select country,year,biofuel_consumption as current_biofuel
from df
where Iso_code is not null and biofuel_consumption is not null  and year>=2015)
select t1.country,
t1.year as current_year,
round(t1.current_biofuel,2)as biofuel_this_year,
round(t2.current_biofuel,2) as biofuel_last_year,
round(((t1.current_biofuel-t2.current_biofuel)/t2.current_biofuel)*100,2)as YOY_biofuel_growth_percentage
from annual_biofuel t1
join annual_biofuel t2
    on t1.country=t2.country
    and t1.year=t2.year+1
    where t1.country='India'
    order by current_year DESC;
    """).show()
```

## current year solar_global_average

```python
duckdb.sql("""
select main.year,round(main.solar_consumption,2)as India_solar,
(select round(avg(sub.solar_consumption),2)
from df sub
where sub.year=main.year
and sub.Iso_code is not null)as global_solar_avg
from df main
where main.country='India'
and main.solar_consumption>(
select avg(sub.solar_consumption)
from df sub 
where sub.year = main.year
and sub.Iso_code is not null
)
order by main.year DESC;
""").show()
```

## current year wind_global_average

```python
duckdb.sql("""
select main.year,round(main.wind_consumption,2)as India_wind,
(select round(avg(sub.wind_consumption),2)
from df sub
where sub.year=main.year
and sub.Iso_code is not null)as global_wind_avg
from df main
where main.country='India'
and main.wind_consumption>(
select avg(sub.wind_consumption)
from df sub 
where sub.year = main.year
and sub.Iso_code is not null
)
order by main.year DESC;
""").show()
```

# country by country solar_global_avg

```python
duckdb.sql("""
WITH global_average AS (
    SELECT AVG(solar_consumption) AS true_avg
    FROM df
    WHERE year = 2023
      AND iso_code IS NOT NULL
      AND LENGTH(iso_code) = 3
      AND country NOT IN ('World', 'Asia', 'Europe', 'North America', 'European Union', 'High-income countries')
)
SELECT 
    main.country,
    main.year,
    ROUND(main.solar_consumption, 2) AS country_solar
FROM df main, global_average
WHERE main.year = 2023
  AND main.iso_code IS NOT NULL
  AND LENGTH(main.iso_code) = 3
  AND main.country NOT IN ('World', 'Asia', 'Europe', 'North America', 'European Union', 'High-income countries')
  AND main.solar_consumption > global_average.true_avg
ORDER BY country_solar DESC
LIMIT 10;
""").show()
```

# country by country Wind_global_avg

```python
duckdb.sql("""
WITH global_average AS (
    SELECT AVG(wind_consumption) AS true_avg
    FROM df
    WHERE year = 2023
      AND iso_code IS NOT NULL
      AND LENGTH(iso_code) = 3
      AND country NOT IN ('World', 'Asia', 'Europe', 'North America', 'European Union', 'High-income countries')
)
SELECT 
    main.country,
    main.year,
    ROUND(main.wind_consumption, 2) AS country_wind
FROM df main, global_average
WHERE main.year = 2023
  AND main.iso_code IS NOT NULL
  AND LENGTH(main.iso_code) = 3
  AND main.country NOT IN ('World', 'Asia', 'Europe', 'North America', 'European Union', 'High-income countries')
  AND main.wind_consumption > global_average.true_avg
ORDER BY country_wind DESC
LIMIT 10;
""").show()
```
